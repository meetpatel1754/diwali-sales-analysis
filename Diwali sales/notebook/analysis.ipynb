## 10.1

```python
gender_sales = df.groupby("Gender")["Amount"].sum().sort_values(ascending=False)

gender_sales.plot(kind="bar", figsize=(7,5))
plt.title("Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")
plt.show()
```

## 10.2

```python
state_sales = df.groupby("State")["Amount"].sum().sort_values(ascending=False)

state_sales.head(10).plot(kind="bar", figsize=(10,5))
plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

## 10.3

```python
category_sales = df.groupby("Product_Category")["Amount"].sum().sort_values(ascending=False)

category_sales.plot(kind="bar", figsize=(10,5))
plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
gender_category = pd.pivot_table(
    df,
    values="Amount",
    index="Product_Category",
    columns="Gender",
    aggfunc="sum"
)

gender_category
```

```python
age_category = pd.pivot_table(
    df,
    values="Amount",
    index="Product_Category",
    columns="Age Group",
    aggfunc="sum"
)

age_category
```

## 11. Key Findings / Business Insights

### Key Finding 1 – Gender
Female customers contributed a higher share of total revenue than male customers.

### Key Finding 2 – Age Group
The highest revenue contribution came from the leading age group identified in the analysis.

### Key Finding 3 – Geography
The top-performing states contributed a significant portion of total sales.

### Key Finding 4 – Product Category
The highest-selling product categories generated the majority of product revenue.

### Key Finding 5 – Occupation
Customers from the leading occupation groups contributed more revenue than other occupation groups.

## 12. Business Recommendations

### Recommendation 1 – Focus on High-Value Customer Segments

The company should focus marketing campaigns on the customer segments that contribute the highest revenue. Personalized offers, loyalty rewards, and targeted promotions can be used to improve repeat purchases.

### Recommendation 2 – Focus on Top-Performing States

The company should prioritize its marketing and promotional activities in the highest-performing states. These regions can be given more targeted campaigns and product availability to increase sales further.

### Recommendation 3 – Promote High-Performing Product Categories

The company should maintain sufficient inventory of the top-performing product categories and promote them through targeted discounts and campaigns.

### Recommendation 4 – Use Cross-Selling Opportunities

The company can use customer and product-category patterns to recommend related products. This can help increase the average order value and overall revenue.

### Recommendation 5 – Review Low-Performing Categories

Low-performing product categories should be investigated to understand whether the issue is related to pricing, demand, product availability, or marketing. Based on the findings, the company can improve, reposition, or reduce these products.

### Recommendation 6 – Use Data-Driven Marketing

Marketing campaigns should be targeted using customer demographics, occupation, location, and purchasing behaviour instead of using the same campaign for all customers.
